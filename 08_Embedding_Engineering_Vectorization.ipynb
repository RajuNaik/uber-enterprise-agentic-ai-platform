# ⚙️ Environment & Embedding Configuration Initialization

```python
# ==========================================
# Environment & Embedding Configuration
# ==========================================

from pyspark.sql.functions import *
from pyspark.sql.types import *

import pandas as pd
import numpy as np

# Enterprise Configuration
CONFIG = {

    "catalog": "spark_catalog",
    "schema": "uber_ai",

    "environment": "dev"
}

# Embedding Configuration
EMBEDDING_CONFIG = {

    # Embedding Model
    "embedding_model":
        "sentence-transformers/all-MiniLM-L6-v2",

    # Embedding Dimensions
    "embedding_dimension":
        384,

    # Batch Size
    "embedding_batch_size":
        32
}

print("✅ Embedding Configuration Initialized")
print(EMBEDDING_CONFIG)
```

# 🥈 Read Enterprise Semantic Chunks

```python
# ==========================================
# Read Enterprise Semantic Chunks
# ==========================================

semantic_chunks_df = spark.table(
    f"{CONFIG['schema']}.semantic_chunks"
)

print("✅ semantic_chunks loaded")

display(
    semantic_chunks_df.limit(5)
)
```

# 🧬 Initialize Enterprise Embedding Model

```python
%pip install sentence-transformers
```

```python
# ==========================================
# Install Sentence Transformers
# ==========================================

# %pip install sentence-transformers

# ==========================================
# Import Embedding Libraries
# ==========================================

from sentence_transformers import SentenceTransformer

# ==========================================
# Load Enterprise Embedding Model
# ==========================================

embedding_model = SentenceTransformer(
    
    EMBEDDING_CONFIG["embedding_model"]
)

print("✅ Enterprise Embedding Model Loaded")

print(
    "Embedding Model:",
    EMBEDDING_CONFIG["embedding_model"]
)

print(
    "Embedding Dimension:",
    EMBEDDING_CONFIG["embedding_dimension"]
)
```

# Enterprise Embedding Pipeline

↓

↓

```python
# 🧬 Enterprise Embedding Vector Generation
```

```python
# ==========================================
# Enterprise Embedding Vector Generation
# ==========================================

# ------------------------------------------
# Read Small Batch for Embedding Generation
# ------------------------------------------

embedding_input_pd = (

    semantic_chunks_df

    .select(
        "chunk_id",
        "document_id",
        "chunk_text",
        "semantic_domain",
        "city_name",
        "retrieval_priority",
        "event_date"
    )

    .limit(100)

    .toPandas()
)

print("✅ Embedding Input Loaded")

# ------------------------------------------
# Generate Embeddings
# ------------------------------------------

embedding_input_pd["embedding_vector"] = (

    embedding_input_pd["chunk_text"]

    .apply(
        lambda x:
            embedding_model.encode(x).tolist()
    )
)

print("✅ Embedding Vectors Generated")

# ------------------------------------------
# Embedding Metadata
# ------------------------------------------

embedding_input_pd["embedding_dimension"] = (

    embedding_input_pd["embedding_vector"]

    .apply(len)
)

embedding_input_pd["embedding_model"] = (
    
    EMBEDDING_CONFIG["embedding_model"]
)

# ------------------------------------------
# Convert Back to Spark DataFrame
# ------------------------------------------

embedding_vectors_df = spark.createDataFrame(
    embedding_input_pd
)

print("✅ Enterprise Embedding Vector Registry Created")

display(
    embedding_vectors_df.limit(5)
)
```

# 💾 Persist Enterprise Vector Registry

```python
# ==========================================
# Persist Enterprise Vector Registry
# ==========================================

(
    embedding_vectors_df.write

    .format("delta")

    .mode("overwrite")

    .saveAsTable(
        f"{CONFIG['schema']}.embedding_vectors"
    )
)

print("✅ embedding_vectors table persisted successfully")
```
